# 구구단 프로그램 (Google Colab 실습)

구구단은 **서랍장**이라고 생각하면 쉽습니다.
서랍장이 8개(2단~9단) 있고, 각 서랍장마다 칸이 9개(x1 ~ x9) 있어요.
그래서 반복문이 두 겹이 됩니다. 바깥쪽이 *어느 서랍장*, 안쪽이 *몇 번째 칸*을 고릅니다.

아래 셀을 **위에서 아래로 하나씩** 실행해 보세요. (`Shift + Enter`)


## 1단계 — 한 개의 단만 출력하기

먼저 서랍장 하나(7단)만 열어봅니다. 반복문 한 겹이면 충분해요.


````python
dan = 7

for i in range(1, 10):
    print(f"{dan} x {i} = {dan * i}")
````

### 여기서 꼭 기억할 두 가지

**1. `range(1, 10)`은 10을 포함하지 않습니다.**
시작은 포함, 끝은 제외예요. 9까지 나오게 하려고 10이라고 씁니다.
초보자가 가장 자주 틀리는 지점입니다. 아래 셀로 직접 확인해 보세요.

**2. f-string** — 문자열 앞에 `f`를 붙이면 `{}` 안의 값이 계산되어 끼워집니다.
`{dan * i}`처럼 계산식도 바로 넣을 수 있어요.


````python
print(list(range(1, 10)))   # 1부터 9까지. 10은 없다
print(list(range(2, 10)))   # 2부터 9까지
print(list(range(10)))      # 시작을 안 쓰면 0부터
````

## 2단계 — 2단부터 9단까지 전부 출력하기

서랍장을 하나씩 바꿔가며 1단계를 반복하면 됩니다. 반복문을 두 겹으로 겹칩니다.

```
for dan in ...      <- 바깥: 어느 서랍장을 열까 (2~9)
    for i in ...    <- 안쪽: 그 서랍장의 1~9번 칸
```

**들여쓰기(indent)가 곧 문법입니다.** 안쪽 `for`가 바깥 `for`보다 한 칸(보통 4칸) 더
들어가 있어야 "바깥 반복문에 속한 코드"가 됩니다. 파이썬은 중괄호 대신 들여쓰기로
코드 덩어리를 구분해요.


````python
for dan in range(2, 10):
    print(f"--- {dan}단 ---")
    for i in range(1, 10):
        print(f"{dan} x {i} = {dan * i}")
    print()   # 단 사이에 빈 줄 하나
````

## 3단계 — 함수로 묶기

같은 일을 여러 번 시킬 거라면 **함수**로 이름을 붙여둡니다.
요리에 비유하면, 매번 재료 손질부터 설명하는 대신 "김치찌개 끓이기"라는
이름표를 붙여두고 필요할 때 그 이름만 부르는 것과 같아요.

함수를 정의하는 셀을 실행하면 **아무 출력도 안 나옵니다.** 정상입니다.
"이름표를 등록"만 한 것이고, 실제로 일하는 건 아래에서 부를 때예요.


````python
def print_one_dan(dan):
    """한 개의 단(예: 7단)을 출력한다."""
    print(f"--- {dan}단 ---")
    for i in range(1, 10):
        print(f"{dan} x {i} = {dan * i}")


def print_all():
    """2단부터 9단까지 차례로 출력한다."""
    for dan in range(2, 10):
        print_one_dan(dan)
        print()

print("함수 등록 완료")
````

이제 부르기만 하면 됩니다. 숫자를 바꿔서 여러 번 실행해 보세요.


````python
print_one_dan(3)
print()
print_one_dan(8)
````

````python
print_all()   # 2단 ~ 9단 전부
````

## 4단계 — 사용자가 직접 숫자를 넣게 하기

`input()`을 쓰면 코랩 화면에 입력창이 나타납니다. 숫자를 치고 `Enter`를 누르세요.

주의할 점이 하나 있습니다. **`input()`이 돌려주는 값은 항상 문자열입니다.**
`"7"`은 글자이고 `7`은 숫자예요. 글자끼리는 곱셈이 안 되니 `int()`로 바꿔줘야 합니다.

그리고 사용자가 `abc`처럼 엉뚱한 값을 넣으면 `int()`가 에러를 냅니다.
그래서 `try / except`로 감싸 "에러가 나면 프로그램이 죽는 대신 안내 문구를 보여주도록" 합니다.
안전벨트 같은 장치예요.


````python
value = input("몇 단을 볼까요? (1~9): ")

try:
    dan = int(value)          # 문자열 -> 숫자
except ValueError:
    print(f"숫자를 입력해 주세요. 받은 값: {value}")
else:
    if 1 <= dan <= 9:
        print_one_dan(dan)
    else:
        print(f"1부터 9 사이의 숫자만 됩니다. 받은 값: {dan}")
````

## 5단계 (보너스) — 표로 예쁘게 보기

구구단은 사실 가로 9칸 x 세로 9칸짜리 **표**입니다.
코랩에는 `pandas`가 이미 깔려 있어서 표를 바로 그릴 수 있어요.

`[[i * j for j in ...] for i in ...]` 형태를 **리스트 컴프리헨션**이라고 부릅니다.
`for` 문을 한 줄로 압축한 문법이고, 결과는 "리스트 안에 리스트"(= 표)가 됩니다.


````python
import pandas as pd

numbers = range(1, 10)
table = [[i * j for j in numbers] for i in numbers]

df = pd.DataFrame(table, index=numbers, columns=numbers)
df.index.name = "단"
df.columns.name = "x"
df
````
